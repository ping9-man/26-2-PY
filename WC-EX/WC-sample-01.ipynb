```python
from wordcloud import WordCloud, STOPWORDS
from matplotlib import font_manager, rc
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
import platform

# 주피터 노트북 환경이 아닐 경우 %matplotlib inline은 주석 처리하세요.
# %matplotlib inline

# 1. 파일 경로 (Raw 문자열 사용)
text = open(r'C:\Users\Lucky\2026-2 PY\WC-EX\obama-speech-EN.txt', 'r', encoding='utf-8').read()
jesus_mask = np.array(Image.open(r'C:\Users\Lucky\2026-2 PY\WC-EX\image_04.png'))

stopwords = set(STOPWORDS)
stopwords.add("unto")
stopwords.add('ye')

path = r'C:\Windows\fonts\malgun.ttf'
font_name = font_manager.FontProperties(fname=path).get_name()
rc('font', family=font_name)

# mask 이미지 확인용 출력 (출력 화면 비율 조정)
plt.figure(figsize=(8.40, 6.40))
plt.imshow(jesus_mask, cmap=plt.cm.gray, interpolation='bilinear')
plt.axis('off')
plt.show()

# 2. 워드클라우드 크기를 1024x768로 지정
wc = WordCloud(
    width=860,          # 가로 크기 지정
    height=640,          # 세로 크기 지정
    background_color='white', 
    max_words=2000, 
    mask=jesus_mask,
    stopwords=stopwords
)
wc = wc.generate(text)

# 단어 빈도수 확인 (선택 사항)
# print(wc.words_)

# 3. 최종 결과 출력 (1024x768 비율에 맞춘 시각화)
plt.figure(figsize=(8.60, 6.40))
plt.imshow(wc.to_array(), interpolation='bilinear')
plt.axis('off')
plt.show()
```
